```python
import asyncio
import time
```

```python
# Synchronus approach
def write():
    print("Hey")
    time.sleep(1)
    print("there")

def main():
    for i in range(10):
        write()

if __name__=="__main__":
    start=time.perf_counter()
    main()
    end=time.perf_counter()-start
    print(f"Executed in {end:0.2f} seconds.")
```

## **Solution:** *AsyncIO(Reducing Execution Time)*
- In Python as we mark a function with `async` , it becomes **coroutine** and *coroutines can be paused and resumed , allowing other tasks to run in meantime.*
- By using `async-await` we can write non-blocking code that can perform tasks concurrently, resulting **Faster** and **More responsive** programs.

```python
# asynchronus Approach
async def write():
    print("Hey")
    await asyncio.sleep(5)
    print("there")

async def main():
    await asyncio.gather(write(),write(),
                         write(),write(),
                         write(),write(),
                         write(),write(),
                         write(),write(),
                         )

if __name__=="__main__":
    start=time.perf_counter()
    await main()
    end=time.perf_counter()-start
    print(f"Executed in {end:0.2f} seconds.")
```

## **Conclusion**
- Thus we see that *total execution time is decrease*d significantly.
- so we can use async-await in scenarios like if we are loading data in meantime we can perform another task not relevant to data being loaded and it will give better performance.
